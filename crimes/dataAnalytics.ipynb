```python
import pandas as pd
import numpy as np
import calendar
import seaborn as sns
import matplotlib.pyplot as plt
import folium
from folium.plugins import HeatMap
import warnings
warnings.filterwarnings('ignore')
```

<hr>
<h1>CHICAGO'S CRIME</h1>

```python
crimes = pd.read_csv('datasets//Chicago_Crimes.csv')
crimes
```

<hr>
<h2>CLEANING DATA</h2>

```python
crimes.isnull().sum()
```

```python
crimes['Location Description'] = crimes['Location Description'].fillna(crimes['Location Description'].mode()[0])
crimes['Community Area'] = crimes['Community Area'].fillna(crimes['Community Area'].mode()[0])
crimes['X Coordinate'] = crimes['X Coordinate'].fillna(crimes['Latitude'].mode()[0])
crimes['Y Coordinate'] = crimes['Y Coordinate'].fillna(crimes['Latitude'].mode()[0])
crimes['Latitude'] = crimes['Latitude'].fillna(crimes['Latitude'].mode()[0])
crimes['Longitude'] = crimes['Longitude'].fillna(crimes['Longitude'].mode()[0])
crimes['Location'] = crimes['Location'].fillna(crimes['Latitude'].mode()[0])
```

```python
crimes.isnull().sum()
```

<hr>
<h2>CHANGING DATA TYPES</h2>

```python
crimes.dtypes
```

```python
crimes['Date'] = crimes['Date'].str.strip()

crimes['Date'] = crimes['Date'].astype('str')

crimes['Date'] = crimes['Date'].str.replace('/', '-')

crimes['Date'] = pd.to_datetime(crimes['Date'], dayfirst=True, errors='coerce')

# for loop para sa object to category, tinamad ako isa isahinTT_TT
for x in crimes.select_dtypes(include='object').columns:
    crimes[x] = crimes[x].astype('category')

crimes['Year'] = crimes['Date'].dt.year
crimes['Month'] = crimes['Date'].dt.month
crimes['Day'] = crimes['Date'].dt.day
crimes['DayOfWeek'] = crimes['Date'].dt.dayofweek
```

```python
crimes.dtypes
```

<hr>
<h2>UNIVARIATE ANALYSIS</h2>
<h4>1. What are the top 5 most common crime types?</h4>

```python
topFiveCrimes = crimes['Primary Type'].value_counts().head(5).index

topCrimes = crimes[crimes['Primary Type'].isin(topFiveCrimes)]

plt.figure(figsize=(10, 6))
graph = sns.countplot(
    y='Primary Type',
    data=topCrimes,
    order=topFiveCrimes,
    color='skyblue'
)

for x in graph.patches:
    count = int(x.get_width())
    graph.text(count + 500,
            x.get_y() + x.get_height() / 2,
            f'{count:,}',
            va='center', fontsize=11)

plt.title("Top 5 Crime Types in Chicago")
sns.despine(left=True, bottom=True)
plt.tight_layout()
plt.show()
```

<h4>INSIGHTS #1</h4>
<p>The top 5 most common crime types in Chicago are Theft, Battery, Criminal Damage, Assault and Motor Vehicle Theft</p>

<h4>INSIGHTS #2</h4>
<p><b>Theft</b> is the most common crime in Chicago, with 59,201 recorded cases, far higher than any other crime type.</p>

<h4>INSIGHTS #3</h4>
<p><b>Battery</b> ranks second with 44,579 cases, which is also a significant problem compared to other crimes.</p>

<h4>INSIGHTS #4</h4>
<p><b>Criminal damage</b> (27,094 cases), while still common, occurs much less frequently than theft and battery.</p>

<h4>INSIGHTS #5</h4>
<p><b>Assault (22,615 cases) and motor vehicle theft (19,604 cases)</b> are the least frequent among the top five, but they still represent a large number of incidents.</p>

<h4>2. What is the distribution of crimes by month?</h4>

```python
monthNames = list(calendar.month_name)[1:]

plt.figure(figsize=(10, 6))
graph = sns.countplot(
    x='Month',
    data=crimes,
    order=range(1, 13),
    color='skyblue'
)

for x in graph.patches:
    height = x.get_height()
    graph.text(
        x.get_x() + x.get_width() / 2,
        height + 10,
        f'{int(height):,}',
        ha='center',
        fontsize=11
    )

plt.xticks(ticks=range(12), labels=monthNames, rotation=45)
plt.title("Crimes by Month")
sns.despine()
plt.tight_layout()
plt.show()
```

<h4>INSIGHTS #6</h4>
<p><b>July</b> has the highest number of crimes (22,957 cases), showing that crime peaks in the summer.</p>

<h4>INSIGHTS #7</h4>
<p><b>February</b> records the lowest number of crimes (17,393 cases), which may be influenced by the shorter month and colder weather.</p>

<h4>INSIGHTS #8</h4>
<p>From <b>May to October</b>, crime levels are consistently high (all above 21,000), suggesting that warmer months generally see more crime.</p>

<h4>INSIGHTS #9</h4>
<p>The overall trend suggests a seasonal pattern, with crime rising in spring and peaking in summer, then declining towards winter.</p>

<h4>13. Yearly crime trend by weekday and arrest</h4>

```python
plt.figure(figsize=(12, 6))

sns.lineplot(
    x='Year', 
    y='Arrest', 
    hue='DayOfWeek', 
    data=crimes, 
    estimator=np.mean,
    errorbar=None
)

plt.title("Arrest Probability by Day of Week Over Years")
plt.xlabel("Year")
plt.ylabel("Arrest Probability")

plt.xticks(sorted(crimes['Year'].unique()))
plt.tight_layout()
plt.show()
```

<h4>INSIGHTS #31</h4>
<p>Sunday and Monday consistently show higher arrest probabilities compared to others such as Friday and Saturday. This suggests that the day of the week influences policing patterns or crime reporting/arrest likelihood.</p>

<h4>INSIGHTS #32</h4>
<p>The analysis reveals that arrest probabilities have increased steadily across all days of the week from 2024 to 2025, with certain days (especially earlier in the week) having consistently higher arrest rates. This suggests that law enforcement effectiveness is improving over time, though variations by weekday indicate that crime patterns and arrest likelihood are not uniform throughout the week.</p>

<h4>Multivariate correlation HeatMap</h4>

```python
plt.figure(figsize=(10,6))
sns.heatmap(crimes[['Arrest','Domestic','Year','Month','Hour']].corr(), annot=True, cmap="coolwarm")
plt.title("Correlation HeatMap of Crime Variables")
plt.show()
```

<h4>INSIGHTS #34</h4>
<p>Year vs. Month has the strongest negative correlation (-0.42). This makes sense because "Month" repeats every year while "Year" keeps increasing, leading to an inverse relationship in the dataset encoding rather than a real-world crime trend.</p>

<h4>INSIGHTS #35</h4>
<p>Arrest vs. Year has a slight positive correlation (0.05). Indicating that in more recent years, there might be a trend of slightly higher arrest rates, consistent with earlier trends you saw in the line plot.</p>

<h4>INSIGHTS #36</h4>
<p>Arrest vs. Hour shows a slight positive correlation (0.052). This suggests that the time of day has some small influence on whether an arrest is made (possibly due to higher police activity during certain hours).</p>

<h4>INSIGHTS #37</h4>
<p>Weak correlations overall, most variables show correlations very close to zero, meaning that factors like arrest, domestic crimes, year, month, and hour are largely independent of each other in this dataset.</p>
